```python
# Riddhi Mehrotra
# Assignment 1

# a. Create a list of numbers and print all even numbers

n = int(input("Enter the number of elements: "))

numbers = []

for i in range(n):
    num = int(input("Enter a number: "))
    numbers.append(num)

print("List:", numbers)

print("Even numbers:")
for num in numbers:
    if num % 2 == 0:
        print(num, end=" ")
```

```python
# b. Create a string and print it in reverse order.
# Also reverse the words.

text = input("\n\nEnter a string: ")

# Reverse the complete string
reverse_string = text[::-1]

print("Reversed string:")
print(reverse_string)

# Reverse the order of words
words = text.split()
reverse_words = words[::-1]

print("Words in reverse order:")
print(" ".join(reverse_words))
```
